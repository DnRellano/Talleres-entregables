```python
import requests
import pandas as pd

base_url = "https://api.deezer.com/chart"
endpoint = ""
response = requests.get(base_url + endpoint)

if response.status_code == 200:
  print("Conexión exitosa a la API. Codigo de estado:", response.status_code)
  data = response.json() # Parse the JSON response
else:
  print("Conexión fallida a la API. Codigo de estado", response.status_code)
```

```python
if 'data' in globals() and 'tracks' in data and 'data' in data['tracks']:
  print("Displaying the chart tracks:")
  df_tracks = pd.DataFrame(data['tracks']['data'])
  display(df_tracks)
else:
  print("Data not available or 'tracks' data not found. Please ensure the API call in the previous cell was successful and returned data, and that 'tracks' data exists.")
```

```python
# Columnas a eliminar
columns_to_drop = ['title_version', 'link', 'explicit_content_lyrics', 'explicit_content_cover', 'preview', 'md5_image']

# Eliminar las columnas del DataFrame
df_tracks_cleaned = df_tracks.drop(columns=columns_to_drop)

# Extraer el nombre del artista de la columna 'artist'
df_tracks_cleaned['artist_name'] = df_tracks_cleaned['artist'].apply(lambda x: x['name'] if isinstance(x, dict) and 'name' in x else None)

# Extraer el título del álbum de la columna 'album'
df_tracks_cleaned['album_title'] = df_tracks_cleaned['album'].apply(lambda x: x['title'] if isinstance(x, dict) and 'title' in x else None)

# Eliminar las columnas originales 'artist' y 'album' (que contenían diccionarios)
df_tracks_cleaned = df_tracks_cleaned.drop(columns=['artist', 'album'])

# Mostrar las primeras filas del DataFrame modificado
print("DataFrame después de eliminar columnas y extraer nombres de artista y álbum:")
display(df_tracks_cleaned.head())

# Guardar el DataFrame limpio a un nuevo archivo CSV (o sobrescribir el anterior si se desea)
csv_filename_cleaned = 'deezer_tracks_cleaned.csv'
df_tracks_cleaned.to_csv(csv_filename_cleaned, index=False)

print(f"\nLos datos limpios se han guardado exitosamente en '{csv_filename_cleaned}'")

#### SE ACLARA QUE CADA SEMANA CAMBIA EL RANKIN DE CANCIONES Y ARTISTAS, NO SIEMPRE ES LA MISMA, POR ENDE TOCA ACTUALIZAR EL DASHBOARD
```
